## Step 1: Import Required Libraries

## Step 2: Load the Dataset

The dataset is stored inside the `data` folder.

```python
df = pd.read_csv("../data/winequality-red.csv")
df.head()
```

# Visualization 1: Histogram

### Purpose

A histogram shows how values are distributed across different ranges.

Here we visualize the **Alcohol** column.

```python
plt.figure(figsize=(8,5))

plt.hist(df["alcohol"], bins=15, edgecolor="black")

plt.title("Distribution of Alcohol Content")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")

plt.show()
```

### Observation

Most wine samples have alcohol values concentrated around the middle range, while higher alcohol values appear less frequently.

# Visualization 2: Boxplot

### Purpose

A boxplot helps identify outliers and understand the spread of data using quartiles.

```python
plt.figure(figsize=(8,3))

plt.boxplot(df["alcohol"], vert=False)

plt.title("Boxplot of Alcohol Content")
plt.xlabel("Alcohol")

plt.show()
```

### Observation

The boxplot highlights the median, quartiles, and any unusually high alcohol values that behave as outliers.

# Visualization 3: Scatter Plot

### Purpose

A scatter plot helps us understand the relationship between two numerical variables.

We will compare:

- Alcohol
- Sulphates

```python
plt.figure(figsize=(8,5))

plt.scatter(df["alcohol"], df["sulphates"])

plt.title("Alcohol vs Sulphates")
plt.xlabel("Alcohol")
plt.ylabel("Sulphates")

plt.show()
```

### Observation

The scatter plot suggests that sulphate levels vary with alcohol content, although the relationship is not perfectly linear.

# Visualization 4: Correlation Heatmap

### Purpose

A heatmap helps identify strong positive and negative relationships among numerical features.

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

The heatmap clearly highlights which chemical properties have stronger positive or negative relationships with each other.

# Strongest Correlation

Now let's identify the strongest correlated feature pairs after removing self-correlation.

```python
corr = df.corr(numeric_only=True)

corr_pairs = (
    corr.where(corr != 1)
        .stack()
        .sort_values(key=abs, ascending=False)
)

corr_pairs.head()
```
